```python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    result = num1 / num2
except ZeroDivisionError:
    print("Error: Cannot divide by zero")
except ValueError:
    print("Error: Invalid input")
else:
    print("Result =", result)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        self._age = value

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

p = Person("Raghad", 20)
p.age = 21
p.display()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Manager(Employee):
    def add_bonus(self, bonus):
        self.salary += bonus

    def display(self):
        print("Name:", self.name)
        print("Total Salary:", self.salary)

m = Manager("Raghad", 5000)
m.add_bonus(1000)
m.display()
```

```python
class Student:
    def __init__(self):
        self.name = input("Enter name: ")
        try:
            age_input = int(input("Enter age: "))
            if age_input < 0:
                raise ValueError
            self.age = age_input
        except ValueError:
            print("Error: Invalid age")
            self.age = None

    def display(self):
        if self.age is not None:
            print("Name:", self.name)
            print("Age:", self.age)

s = Student()
s.display()
```
