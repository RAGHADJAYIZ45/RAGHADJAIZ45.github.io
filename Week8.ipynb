```python
#451005003
import os
import shutil

main_folder = "lab_data"

if not os.path.exists(main_folder):
    os.mkdir(main_folder)
    print("Creates a folder named lab_data")
else:
    print("Raghad")

subfolders = ["input", "output", "logs"]

for folder in subfolders:
    path = os.path.join(main_folder, folder)
    if not os.path.exists(path):
        os.mkdir(path)

print("List of lab_data contents:")
for item in os.listdir(main_folder):
    full_path = os.path.join(main_folder, item)
    
    if os.path.isdir(full_path):
        print(item, "Directory")
    else:
        print(item, "File")

old_name = os.path.join(main_folder, "logs")
new_name = os.path.join(main_folder, "history")

if os.path.exists(old_name):
    os.rename(old_name, new_name)
    print("Rename logs to history")
else:
    print("Raghad Jayez")
```

```python
#451005003
notes_path = os.path.join(main_folder, "input", "notes.txt")

with open(notes_path, "w") as file:
    file.write("Line 1\n")
    file.write("Line 2\n")
    file.write("Line 3\n")
    file.write("Line 4\n")
    file.write("Line 5\n")

with open(notes_path, "r") as file:
    lines = file.readlines()
    print("Total number of lines:", len(lines))

summary_path = os.path.join(main_folder, "output", "summary.txt")

with open(summary_path, "w") as summary:
    summary.writelines(lines)

print("Copy the contents to summary.txt")
```
