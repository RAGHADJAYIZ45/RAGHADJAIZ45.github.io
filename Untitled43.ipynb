```python
import random

secret = random.randint(1, 20)
count = 0

while True:
    guess = int(input("Guess the number (1-20): "))

    if guess < 1 or guess > 20:
        print("Invalid number!")
        continue

    count += 1

    if guess == secret:
        print("Correct!")
        print("Attempts:", count)
        break
    else:
        print("Try again")
```
