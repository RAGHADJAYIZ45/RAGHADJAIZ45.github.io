```python
import re

def sum_numbers(numbers):
    return sum(numbers)

text = input("Enter a string: ")

numbers = list(map(int, re.findall(r'\d+', text)))

total = sum_numbers(numbers)

double = lambda x: x * 2

doubled_value = double(total)

print("Numbers found:", numbers)
print("Sum:", total)
print("Doubled value:", doubled_value)
```
