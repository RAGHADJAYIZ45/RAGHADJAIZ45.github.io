```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("bank_customers_time.xlsx")

df.to_excel("bank_customers_time.xlsx", index=False)

print(df.head(10))

print(df.tail(10))

print(df.shape)

print(df.dtypes)

print(df.isnull().sum())

df = df.dropna()

print(df['Balance'].mean())

print(df['Balance'].median())

print(df['Balance'].std())

print(df['Balance'].min())

print(df['Balance'].max())

plt.hist(df['Balance'], bins=20)
plt.title("Bank Balance Histogram")
plt.xlabel("Balance")
plt.ylabel("Frequency")
plt.savefig("hist.png")
plt.show()

df["Income"].value_counts().head(5).plot(kind="bar")
plt.title("Top Jobs")
plt.xlabel("Job")
plt.ylabel("Count")
plt.savefig("bar.png")
plt.show()

df['Balance'].head(20).plot(kind="line")
plt.title("Balance Line Chart")
plt.xlabel("Index")
plt.ylabel("Balance")
plt.savefig("line.png")
plt.show()
```
