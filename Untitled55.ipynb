```python
import re
import math

text = input("Enter a paragraph of text: ")

vowel_words = re.findall(r'\b[AEIOUaeiou]\w*', text)

def count_vowels(paragraph):
    vowels = re.findall(r'[AEIOUaeiou]', paragraph)
    return len(vowels)

total_vowels = count_vowels(text)

length_lambda = lambda word: len(word)

word_lengths = list(map(length_lambda, vowel_words))

if len(word_lengths) > 0:
    average_length = round(sum(word_lengths) / len(word_lengths), 2)
else:
    average_length = 0.00

print("Words starting with vowels:", vowel_words)
print("Total vowels in paragraph:", total_vowels)
print("Average vowel-word length:", average_length)
```
